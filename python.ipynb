```python
import numpy as np

# Create a 2x2 array
array_2x2 = np.array([[1, 2], [3, 4]])

print(array_2x2)
```

```python
import numpy as np
# Create a 3x4 array with random values between 0 and 1
random_array = np.random.rand(3, 4)
print(random_array)
```

```python
import numpy as np
# array of ones (3x3)
ones_array = np.ones((3, 3))
print(ones_array)
# array of zeros (2x2)
zeros_array = np.zeros((2, 2))
print(zeros_array)
```

```python
import numpy as np

# Create a 6x3 array with random values
random_array = np.random.rand(6, 3)

# Transpose the array
transposed_array = random_array.T

print("Original 6x3 array:\n", random_array)
print("\nTransposed 3x6 array:\n", transposed_array)
```

```python
import numpy as np
# set dimenisions
n, p, m = 6, 5, 9
# creat random array
A = np.random.rand(n, p) # shape (n, p)
B = np.random.rand(p, m) # shape (p, m)
# compute dot products
result = np.dot(A, B) # shape (n, m)
print("Array A (n, p):\n" , A)
print("\Array B (p, m):\n " , B)
print("\nDot Product (n, m):\n" , result)
```
